# Project 1: Retail Sales Performance Analysis

End-to-end Python notebook covering:
1. Raw CSV ingestion & pandas data cleaning
2. Data quality & reconciliation audit
3. Star-schema export & local SQL verification
4. Executive KPIs, Top 10 products, Monthly trend, and Regional breakdown

```python
from pathlib import Path
import pandas as pd
import sys

BASE_DIR = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(BASE_DIR))

from data_cleaning import read_csv_safely, clean_sales_dataframe, export_star_schema
from analysis import compute_kpis
from dashboard_logic import get_top_products, get_monthly_trend, apply_dashboard_filters

raw_df = read_csv_safely(BASE_DIR / 'dataset' / 'sales_data.csv')
cleaned_df, audit = clean_sales_dataframe(raw_df)
print('Cleaning audit:', audit)
```

```python
# Executive KPIs & Reconciliation
kpis = compute_kpis(cleaned_df)
raw_sales = pd.to_numeric(raw_df['Sales'], errors='coerce').sum()
diff_pct = abs(raw_sales - kpis['Total Revenue']) / abs(raw_sales) * 100

print('KPIs:', kpis)
print(f'Reconciliation diff: {diff_pct:.4f}% (<= 0.1% required)')
```

```python
# Top 10 Products & Monthly Trend
display_cols = ['product name', 'total amount', 'quantity', 'profit']
print('Top 10 Products:')
print(get_top_products(cleaned_df, 10)[display_cols].to_string(index=False))

print('\nMonthly Trend (Last 12 Months):')
print(get_monthly_trend(cleaned_df).tail(12).to_string(index=False))
```

```python
# Test Region Slicer & Category Slicer
central_tech = apply_dashboard_filters(cleaned_df, regions=['Central'], categories=['Technology'])
print('Central + Technology KPIs:', compute_kpis(central_tech))
```
